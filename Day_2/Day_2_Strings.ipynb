### 02 - String Indexing
**Task:** Access individual characters from a string using their index positions. 

```python
word = "Python"
print(word[0])
print(word[-1])
```

### 03 - String Slicing
**Task:** Extract different portions of  a string using slicing.

```python
text = "programming"
print(text[0:6])
print(text[6:])
```

### 04 - Spillting a string
**Task:** Split comma- separated text into sperated values.

```python
csv_line = 'red,green,blue'
print(csv_line.split(','))
```

### 05 - Joining Strings
**Task:** Join Multiple strings together using a serperator.

```python
words = ['Python', 'is', 'fun']
sentence = ' '.join(words)
print(sentence)
```

### 06 - Replacing Text
**Task:** Replace part of a string with new text.

```python
sentence ="I like Java"
new_sentence = sentence.replace("java", "Python")
print(new_sentence)
```

### 07 - Checking string Content
**Task:** Check whether a string contains only alphabetic characters or only digits.

```python
text = "Python"
number = "12345"
print(text.isalpha())
print(number.isdigit())
```

### 08 - Capaitalizing & Centering Text
**Task:** capitalize a name and display it centered within a fixed width

```python
name = 'ravi'
print(name.capitalize())
print(name.center(12,'-'))
```

### 09 - Formatting Strings
**Task:** Convert a product name into a slug and diplay its price in a formatting string.

```python
product = 'Laptop Pro'
price = 55000
slug = product.lower().replace(' ', '-')
print(f'{slug}: rs. {price:,})')
```

### 10 - String length & Reversal
**Task:** Find the length of a string and everse it using slicing

```python
text = "Python"
print(len(text))
print(text[::-1])
```
